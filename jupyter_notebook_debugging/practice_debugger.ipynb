```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "./dummy_data.csv"
data = pd.read_csv(DATA_PATH)

data.shape, data.head()
```

```python
# Exercise 1 

def fill_missing_salary(df):
    """Fill missing salaries with each employee's department average."""
    df = df.copy()
    dept_avg = df["salary"].mean()  # <-- inspect this while stepping
    for idx, row in df.iterrows():
        if pd.isna(row["salary"]):
            df.loc[idx, "salary"] = dept_avg
    return df

filled = fill_missing_salary(data)
filled.loc[data["salary"].isna(), ["department", "salary"]]
```

```python
data["salary"] = data["salary"] + 200
```

```python
# Exercise 2 

def years_to_retirement(age, retirement_age=65):
    return retirement_age - age

def add_retirement_column(df):
    df = df.copy()
    results = []
    for idx, row in df.iterrows():
        yrs = years_to_retirement(row["age"])
        results.append(yrs)
    df["years_to_retirement"] = results
    return df

with_retirement = add_retirement_column(data)
with_retirement[["employee_id", "age", "years_to_retirement"]].head()
```

```python
#Exercise 3

def performance_bonus(row):
    """Bonus = base rate (by department) * salary."""
    if row["department"] == "engineering":
        base = 0.10
    elif row["department"] == "sales":
        base = 0.08
    elif row["department"] == "support":
        base = 0.05
    # no branch for "marketing" -- base is never assigned for those rows
    return base * row["salary"] if pd.notna(row["salary"]) else 0

def apply_bonuses(df):
    df = df.copy()
    df["bonus"] = df.apply(performance_bonus, axis=1)
    return df

bonused = apply_bonuses(data)
bonused.head()
```
